```python
from agents import Agent, WebSearchTool, trace, Runner, gen_trace_id, function_tool
from agents.model_settings import ModelSettings
from pydantic import BaseModel
from dotenv import load_dotenv
import asyncio
import os
import sendgrid
from sendgrid.helpers.mail import Mail
from typing import Dict
from IPython.display import display, Markdown
```

```python
# load environment variables
load_dotenv(override=True)
```

```python
INSTRUCTIONS = """
Search the web for available grants from government of Canada. Provide links and brief descriptions of each grant. Today's date is November 12 2025. Only provide grants that are currently open for applications.
The grants should be relevant to technology startups focused on AI and machine learning. Limit to 3 results.
"""
```

```python
class SearchResult(BaseModel):
    title: str
    link: str
    description: str
```

```python
class Results(BaseModel):
    results: list[SearchResult]
```

```python
search_agent = Agent(
    name="Deep Research Agent",
    instructions=INSTRUCTIONS,
    tools=[WebSearchTool(search_context_size="low")],
    model="gpt-4o-mini",
    model_settings=ModelSettings(tool_choice="required"),
    output_type=Results
)
```

```python
message = "Latest AI and Machine learning grants available to Alberta startups"
```

```python
with trace("Search"):
    result = await Runner.run(search_agent, message)
```

```python
display(Markdown(f"## Search Result\n\n{result.final_output}"))
```
